```python
def pokemon_category(row):
    if row["is_mythical"] == 1:
        return "Mythical"
    elif row["is_legendary"] == 1:
        return "Legendary"
    elif row["is_sublegendary"] == 1:
        return "Sub-Legendary"
    else:
        return "Normal"

df["pokemon_category"] = df.apply(pokemon_category, axis=1)

df["pokemon_category"].value_counts()

category_counts = df["pokemon_category"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

plt.title("Pokémon Classification")
plt.show()
```

```python
against_columns = [
    "against_normal",
    "against_fire",
    "against_water",
    "against_electric",
    "against_grass",
    "against_ice",
    "against_fighting",
    "against_poison",
    "against_ground",
    "against_flying",
    "against_psychic",
    "against_bug",
    "against_rock",
    "against_ghost",
    "against_dragon",
    "against_dark",
    "against_steel",
    "against_fairy"
]

type_effectiveness = df[against_columns].mean()

type_effectiveness

print(type_effectiveness)

type_effectiveness = type_effectiveness.sort_values()

plt.figure(figsize=(12, 7))

plt.barh(
    type_effectiveness.index,
    type_effectiveness.values
)

plt.axvline(
    1,
    linestyle="--"
)

plt.title("Average Defensive Effectiveness Against Each Type")
plt.xlabel("Average Effectiveness")
plt.ylabel("Attack Type")

plt.show()
```

```python
evo_columns = [
    "evochain_0",
    "evochain_2",
    "evochain_4",
    "evochain_6"
]

df["evolution_stages"] = df[evo_columns].notna().sum(axis=1)

df["evolution_stages"].value_counts().sort_index()

def evolution_category(stages):
    if stages == 1:
        return "0 Evolutions"
    elif stages == 2:
        return "1 Evolution"
    elif stages == 3:
        return "2 Evolutions"
    else:
        return "3+ Evolutions"

df["evolution_category"] = df["evolution_stages"].apply(
    evolution_category
)

order = [
    "0 Evolutions",
    "1 Evolution",
    "2 Evolutions",
    "3+ Evolutions"
]

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="evolution_category",
    order=order
)

plt.title("Pokémon by Number of Evolution Stages")
plt.xlabel("Evolution Category")
plt.ylabel("Number of Pokémon")

plt.show()
```

```python
df["has_mega"] = df["mega_evolution"].notna()

df["has_gigantamax"] = df["gigantamax"].notna()

print("Mega Evolution:", df["has_mega"].sum())
print("Gigantamax:", df["has_gigantamax"].sum())

special_forms = pd.Series({
    "Mega Evolution": df["has_mega"].sum(),
    "Gigantamax": df["has_gigantamax"].sum()
})

plt.figure(figsize=(8, 6))

plt.bar(
    special_forms.index,
    special_forms.values
)

plt.title("Pokémon with Special Forms")
plt.xlabel("Special Form")
plt.ylabel("Number of Pokémon")

plt.show()
```
